```python
Amazon['rating_count'] = (
    Amazon['rating_count']
    .astype('string')
    .str.replace(',', '', regex=False)
)

Amazon['rating_count'] = pd.to_numeric(
    Amazon['rating_count'],
    errors='coerce'
)
```

```python
Amazon['rating'].head()
```

```python
Amazon['rating'] = pd.to_numeric(
    Amazon['rating'],
    errors='coerce'
)
```

```python
Amazon.isna().sum()

Amazon['rating_count']=Amazon['rating_count'].fillna(Amazon['rating_count'].median())

```

```python
Amazon.isna().sum()
```

```python
Amazon['rating']=Amazon['rating'].fillna(Amazon['rating'].mean())
```

```python
Amazon.columns
```

```python
Amazon['actual_price'] = (
    Amazon['actual_price']
    .astype('string')
    .str.replace('₹', '', regex=False)
)

Amazon['actual_price'] = pd.to_numeric(
    Amazon['actual_price'],
    errors='coerce'
)
```

```python
Amazon.dtypes
```

```python
Amazon['discounted_price'].head()
Amazon['discounted_price'] = (
    Amazon['discounted_price']
    .astype('string')
    .str.replace('₹', '', regex=False)
)

Amazon['discounted_price'] = pd.to_numeric(
    Amazon['discounted_price'],
    errors='coerce'
)
```

```python
Amazon['discount_percentage'].head()
Amazon['discount_percentage'] = (
    Amazon['discount_percentage']
    .astype('string')
    .str.replace('%', '', regex=False)
)

Amazon['discount_percentage'] = pd.to_numeric(
    Amazon['discount_percentage'],
    errors='coerce'
)
```

```python
Amazon.dtypes
```

```python
Amazon.isna().sum()
```

```python
Amazon['actual_price']=Amazon['actual_price'].fillna(Amazon['actual_price'].mean())
Amazon['discounted_price']=Amazon['discounted_price'].fillna(Amazon['discounted_price'].mean())
```

```python
Amazon.isna().sum()

```

```python
Amazon
```

```python
Amazon['discount_amount'] = (
    Amazon['actual_price'] -
    Amazon['discounted_price']
)
Amazon
```

```python
Amazon['calculated_discount_percentage'] = (
    (Amazon['actual_price'] - Amazon['discounted_price'])
    / Amazon['actual_price']
) * 100
```

```python
Amazon['rating_category'] = pd.cut(
    Amazon['rating'],
    bins=[0, 2, 3, 4, 5],
    labels=['Poor', 'Average', 'Good', 'Excellent']
)
```
